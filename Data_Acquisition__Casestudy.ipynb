```python
import json, sqlite3, time
import numpy as np
import pandas as pd
import requests
import xml.etree.ElementTree as ET
from bs4 import BeautifulSoup
```

Part A – Data Extraction

```python
# Read the Population dataset from the CSV file.
# Hint:
# ● Use the pandas library.
# ● Display the first five rows.
population_df = pd.read_csv('/content/population.csv')
population_df.head()
```

```python
# Read the Population dataset from the CSV file.
# Hint:
# ● Check the shape and data types.
population_df.info()
```

```python
# Read the GDP dataset from the Excel file.
# Hint:
# ● Use the appropriate pandas function.
# ● Display column names and data types.
gdp_df = pd.read_excel('/content/gdp.xlsx')
gdp_df.info()
```

```python
# Read the Internet Users dataset from the JSON file.
# Hint:
# ● Use either the json module or pandas.
# ● Convert the data into a DataFrame.
users_df = pd.read_json('/content/internet_users.json')
users_df.head()
```

```python
# Read the Literacy Rate dataset from the XML file.
# Hint:
# ● Use xml.etree.ElementTree or pandas.read_xml().
# ● Extract all required fields into a DataFrame.
tree = ET.parse('/content/literacy_rate.xml')
root = tree.getroot()

rows = []
for country in root.findall('country'):
    rows.append({
        'name': country.get('name'),
        'literacy_rate': country.findtext('literacy_rate'),
    }) # Collect each item as a dictionary

literacy_rate_df = pd.DataFrame(rows) # Convert the list of dictionaries to a DataFrame
literacy_rate_df.head()
```

Part C – Data Transformation

```python
# Rename columns wherever necessary so that all datasets have a common column name for merging.
gdp_df = gdp_df.rename(columns = {'Country':'country'})
gdp_df.info()
```

```python
# Rename columns wherever necessary so that all datasets have a common column name for merging.
literacy_rate_df = literacy_rate_df.rename(columns = {'name':'country'})
literacy_rate_df.info()
```

```python
# Convert the required columns into appropriate numeric data types.
literacy_rate_df['literacy_rate'] = pd.to_numeric(literacy_rate_df['literacy_rate'])
literacy_rate_df.info()
```

```python
# Merge all four datasets into one DataFrame.
# Hint:
# ● Use the common column (country)
# ● Choose an appropriate join method.
merged_df = (
    population_df.merge(gdp_df, on='country', how='inner')
       .merge(users_df, on='country', how='inner')
       .merge(literacy_rate_df, on='country', how='inner')
)

merged_df.head()
```

```python
# Create the following calculated columns:
# 1. Internet Penetration (%)
# 2. GDP Per Capita
# Hint:
# Round the calculated values to two decimal places.
merged_df["internet_penetration"] = (merged_df["internet_users"] / merged_df["population"] * 100).round(2)
merged_df["gdp_per_capita"] = (merged_df["GDP"] / merged_df["population"]).round(2)

merged_df.head()
```

Part D – Load Data

```python
# Create a SQLite database. Store the final merged dataset into a table named
# ‘country_statistics’
# Hint:
# ● Use the sqlite3 library.
# ● Load the DataFrame using pandas.
# Create database connection
conn = sqlite3.connect("country_statistics.db")
merged_df.to_sql("country_statistics", conn, if_exists="replace", index=False)
```

```python
# Verify that the data has been loaded successfully.
# Hint:
# Retrieve the first five records from the SQLite table.
query = """SELECT * FROM country_statistics LIMIT 5;"""
result = pd.read_sql_query(query, conn)
result
```

Part F – Data Analysis Using SQL

```python
# Display the Top 5 countries by GDP.
query = """SELECT country, gdp FROM country_statistics ORDER BY gdp DESC LIMIT 5;"""
result = pd.read_sql_query(query, conn)
result
```

```python
# Find the average literacy rate.
query = """SELECT AVG(literacy_rate) AS average_literacy_rate FROM country_statistics;"""
result = pd.read_sql_query(query, conn)
result
```

```python
# Display all countries having literacy rate above 90%.
query = """SELECT country, literacy_rate FROM country_statistics WHERE literacy_rate > 90;"""
result = pd.read_sql_query(query, conn)
result
```

```python
# Find the Top 5 countries with the highest internet penetration.
query = """SELECT country, internet_penetration FROM country_statistics ORDER BY internet_penetration DESC LIMIT 5;"""
result = pd.read_sql_query(query, conn)
result
```

```python
# Calculate the total population.
query = """SELECT SUM(population) AS total_population FROM country_statistics;"""
result = pd.read_sql_query(query, conn)
result
```

```python
# Find the country with the highest GDP per capita.
query = """SELECT country, gdp_per_capita FROM country_statistics ORDER BY gdp_per_capita DESC LIMIT 1;"""
result = pd.read_sql_query(query, conn)
result
```
